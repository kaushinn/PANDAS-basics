```python
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

```python
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "species")
```

```python
sns.set_style("dark") 
sns.set_context("notebook") #paper , #poster
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island", palette= "Dark2")
sns.despine()
```

```python
sns.scatterplot(data = penguins, x = "species", y = "body_mass_g", hue = "island", style = "sex", alpha = 0.5)
```

```python
sns.stripplot(data = penguins, x = "species", y = "body_mass_g", hue = "island", dodge= True)
```

```python
sns.stripplot(data = penguins, x = "species", y = "body_mass_g", hue = "island", jitter= True)
```

```python
#swarm plot
sns.set_context("talk")
sns.swarmplot(data = penguins, x = "species", y = "body_mass_g", hue = "island")
sns.despine()
```

```python
sns.set_context("notebook")
sns.histplot(data= penguins, x = "body_mass_g", hue = "sex", multiple = "stack")
```

```python
# regression plot

sns.regplot(data = penguins, x = "flipper_length_mm", y= "body_mass_g", color = "green")
```

```python
# line plot
sns.set_style("darkgrid")
sns.lineplot(data = penguins, x = "flipper_length_mm", y= "body_mass_g", hue ="sex", style= "island")
```

```python
#joint plot

sns.jointplot(data = penguins, x = "flipper_length_mm", y= "body_mass_g", hue ="sex", kind = "hist")
```

```python
sns.barplot(data = penguins, x = "species", y= "body_mass_g", hue ="sex", palette = ["pink", "skyblue"], estimator= np.median)
```

```python
# countplot

sns.countplot(data = penguins, x = "species", hue = "island")
```

```python
# .box pot: 

sns.boxplot(data = penguins, x = "species", y= "body_mass_g", hue ="island")
```

```python
#violen plot
sns.violinplot(data = penguins, x = "species", y= "body_mass_g", hue ="sex", split = True)
```

```python
sns.violinplot(data = penguins, x = "species", y= "body_mass_g")
sns.swarmplot(data = penguins, x = "species", y = "body_mass_g", hue = "island", size = 3, palette = "dark")
```

```python
# kde plot

sns.kdeplot(data = penguins, x = "body_mass_g", hue = "species", palette = "pastel", fill = True)
```
