```python
data = {
    "Salary" : [25000,37000,28000,39000,48000,55000,35000,29000,69000,45000,62000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.shape
```

```python
#line Plot

plt.plot(df["Salary"], color = "red", marker = "*", linestyle = ":", linewidth = "2")
plt.show()
```

```python
#histogram

plt.hist(df["Salary"], bins = 5 , color = "green")
plt.show()
```

```python
#box plot

plt.boxplot(df["Salary"])
plt.show()
```

```python
#univariate Categorical

df["dept"] = ['HR','IT','Finance','IT','HR','Finance','IT','HR','IT','Finance','IT']
```

```python
df.head()
```

```python
df.info()
```

```python
#pie Chart

count = df["dept"].value_counts()
count
```

```python
plt.pie(count,labels = count.index, autopct = "%1.2f", explode = [0,0.1,0.2,0.3])
plt.show()
```

```python
#count plot

plt.bar(count.index, count)
plt.show()
```

```python
new_entry = {"Salary": 98000,"dept": "Engineer"}

df = pd.concat([df, pd.DataFrame([new_entry])], ignore_index = True)
```

```python
print(df)
```

```python
df = df.drop(index=12)
```

```python
#bivariate Analysis - numerical - numerical

df["Age"] = [22,30,23,24,25,36,27,34,20,30,28]
```

```python
df.head()
```

```python
# scatter plot

plt.scatter(df["Age"],df["Salary"], color = "red")
plt.show()
```

```python
sort_age = df.sort_values("Age")
```

```python
#line plot

plt.plot(sort_age["Age"],df["Salary"], color = "red",marker = "*",linestyle= ":", linewidth = "2")
plt.grid()
plt.show()
```

```python
#Bar chart
plt.bar(sort_age["Age"],df["Salary"], color = "red", align = "center")
plt.show()
```

```python
#bivariate numerical - categorical
```

```python
hr_sal = df[df['dept']=="HR"]["Salary"]
hr_sal
```

```python
it_sal = df[df['dept']=="IT"]["Salary"]
it_sal
```

```python
fn_sal = df[df['dept']=="Finance"]["Salary"]
fn_sal
```

```python
er_sal = df[df['dept']=="Engineer"]["Salary"]
er_sal
```

```python
#box plot

plt.boxplot([hr_sal, it_sal, fn_sal, er_sal], tick_labels =["HR","IT","Finance","Engineer"],label =["HR","IT","Finance","Engineer"])
plt.legend()
plt.show()
```

```python
# pie chart

salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels = salary_by_dept.index, autopct = "%1.2f", explode = [0,0.1,0.2], shadow = True)
plt.axis("equal")
plt.show()
```

```python
hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fn_mean = sum(fn_sal)/len(fn_sal)
```

```python
plt.bar(["HR","IT","Finance"] ,[hr_mean, it_mean, fn_mean], color = ["green","black", "red"])
plt.grid()
plt.show()
```

```python
df["Experience"] = [3,5,2,6,8,9,4,5,9,1,6]
```

```python
df.head()
```

```python
# bubble plot

plt.scatter(df["Age"], df["Salary"], s = df["Experience"] , color = "red")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

```python
# 2 Numerical And 1 categorical column
```

```python
color_map = {"HR": "yellow", "IT": "blue", "Finance": "red"}

for dept, group in df.groupby("dept"):
    plt.scatter(
        group["Age"],
        group["Salary"],
        color=color_map.get(dept, "gray"),
        label=dept
    )

plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.title("Age vs Salary vs Dept")
plt.legend(title="Dept")
plt.show()
```

```python
#object oriented Api
```

```python
fig, axs = plt.subplots(1,3, figsize = (15,5))

axs[0].plot(sort_age["Age"], df["Salary"], color = "red", marker = "*", linewidth = "2", markersize = 2)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#histogram

axs[1].hist(df["Salary"], bins = 5, color = "red")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")


#box plot

axs[2].boxplot(df["Salary"])
axs[2].set_title("Box Plot")
axs[2].set_xlabel("Salary")

plt.show()

```

```python
data2 = {
    'Year' : [2020,2021,2022,2023],
    'Sales' : [120,140,110,200],
    'Profit' : [20,30,50,60],
    'Expenses' : [80,120,160,200]
}

    
```

```python
df2 = pd.DataFrame(data2)
```

```python
plt.plot(df2["Year"], df2["Sales"], label = "Sales")
plt.plot(df2["Year"], df2["Profit"], label = "Profit")
plt.plot(df2["Year"], df2["Expenses"], label = "Expenses")

plt.title("Financial Analysis")
plt.xlabel("Year")
plt.ylabel("Amount in $")
plt.legend()
plt.show()
```
